```python
df = pd.read_csv("dataset/imdb.csv")
```

```python
cols_to_drop = [
    "countryOfOrigin",
    "genres",
    "soundMixes",
    "continent",
    "filtered_genres",
    "filtered_soundMixes",
    "Unnamed: 0",
]
df = df.drop(columns=cols_to_drop)
```

```python
TableReport(df)
```

```python
from skrub import tabular_learner
```

```python
target = "rating"
```

```python
X = df.drop(columns=[f"{target}", "averageRating"])
y = df[target]
```

```python
model = tabular_learner("classification").fit(X, y)
report = CrossValidationReport(estimator=model, X=X, y=y)
report.metrics.report_metrics(indicator_favorability=True)
```

```python
report.metrics.accuracy()
```

```python
df2 = pd.read_csv("dataset/train_post_outlier.csv")
```

```python
cols_to_drop = [
    "countryOfOrigin",
    "genres",
    "soundMixes",
    "continent",
    "filtered_genres",
    "filtered_soundMixes",
    "Unnamed: 0",
]
df2 = df2.drop(columns=cols_to_drop)
```

```python
cols_to_drop = ["has_awards", "awardWins", "canHaveEpisodes"]
df2 = df2.drop(columns=cols_to_drop)
```

```python
TableReport(df2)
```

```python
X = df2.drop(columns=["averageRating", "rating_category"])
y = df2["rating_category"]
```

```python
model = tabular_learner("classifier").fit(X, y)
report = EstimatorReport(estimator=model, X_test=X, y_test=y, fit="False")
report.metrics.report_metrics(indicator_favorability=True)
```

```python
report = CrossValidationReport(estimator=model, X=X, y=y, cv_splitter=4)
```

```python
report.metrics.report_metrics(indicator_favorability=True)
```

```python
report.metrics.accuracy()
```
